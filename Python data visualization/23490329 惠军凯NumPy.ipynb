## 【任务2-4】计算同比增速

```python
import numpy as np

last_year_data = np.array([
    [100, 105, 111],
    [115, 113, 120],
    [130, 125, 131],
    [150, 151, 160]])

this_year_data = np.array([
    [110, 125, 135],
    [145, 155, 165],
    [175, 185, 195],
    [205, 215, 225]])
```

```python
# 计算每个月的同比增速（百分比）
growth_rate = (this_year_data - last_year_data) / last_year_data * 100
growth_rate
```

```python
# 计算全年的平均增速（百分比）
average_growth_rate = growth_rate.mean()
average_growth_rate

```

```python
# 计算全年的最高增速（百分比）
max_growth_rate = growth_rate.max()
max_growth_rate

```

```python
# 计算全年的最低增速（百分比）
min_growth_rate = growth_rate.min()
min_growth_rate

```

## 【任务2-5】随机漫步

```python
import numpy as np

# 随机产生的步数情况
steps = 2000
draws = np.random.randint(0, 2, size=steps)
# 当draws的元素为1时，direction_steps的元素为1，
# 当draws的元素为0时，direction_steps的元素为-1
direction_steps = np.where(draws > 0, 1, -1)
direction_steps
```

```python
# 计算步数累计和 （提示：cumsum()函数)
cumulative_steps = direction_steps.cumsum()
cumulative_steps

```

```python
# 计算向前走的最远距离
max_forward_distance = cumulative_steps.max()
max_forward_distance
```

```python
# 计算向后走的最远距离
max_backward_distance = cumulative_steps.min()
max_backward_distance

```

```python
# 将 15 米换算成步数
meters_per_step = 0.8
steps_per_meter = 1 / meters_per_step
steps_for_15_meters = 15 * steps_per_meter
steps_for_15_meters

```

## 【任务2-6】转换销售数据

```python
import numpy as np

# 创建一维数组，代表转换前的销售数据
sales_data = np.array([100, 150, 200, 250, 300, 350])

# 重塑一维数组
sales_data_2d = sales_data.reshape((2, 3))
print("重塑后的二维数组：")
print(sales_data_2d)

# 转置二维数组
sales_data_2d_transposed = sales_data_2d.T
print("转置后的二维数组：")
print(sales_data_2d_transposed)

# 计算每一列的总和
column_sums = sales_data_2d_transposed.sum(axis=0)
print("每一列的总和：")
print(column_sums)

# 将总和添加到二维数组的最后一行
sales_data_2d_with_sums = np.vstack
((sales_data_2d_transposed, column_sums))
print("添加总和后的二维数组：")
print(sales_data_2d_with_sums)

```

## 【任务2-8】兴趣班调查

```python
import numpy as np

# 兴趣班信息
course_data = np.array([
    ["张三", "舞蹈", "钢琴", "武术"],
    ["李四", "绘画", "舞蹈", "游泳"],
    ["王五", "游泳", "舞蹈", "轮滑"],
    ["赵六", "轮滑", "绘画", "武术"],
    ["周七", "外语", "声乐", "绘画"],
    ["韩八", "游泳", "外语", "钢琴"],
    ["秦九", "篮球", "钢琴", "声乐"],
    ["郑十", "轮滑", "篮球", "绘画"]
])

course_data
```

```python
# 提取所有兴趣班（提示：np.flatten
all_courses = course_data[:, 1:].flatten()
print("所有兴趣班：")
print(all_courses)

# 统计各兴趣班的人数（提示：np.unique(x, return_counts=True)
unique_courses, counts = np.unique(all_courses, return_counts=True)
print("兴趣班及其人数：")
for course, count in zip(unique_courses, counts):
    print(f"{course}: {count}人")
```
