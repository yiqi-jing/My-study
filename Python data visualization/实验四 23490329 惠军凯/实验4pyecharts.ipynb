### 快速开始

首先来画出你的第一个 pyecharts 图表

类 ——> 对象 ——> 画图（数据、配置项） ——> 渲染图形

```python
from pyecharts.charts import Bar  # 1. 导入相应的类
from pyecharts.globals import CurrentConfig

CurrentConfig.ONLINE_HOST = "./"  # 切换到运行目录下，去加载 echarts.min.js 文件

# 2. 创建对象
bar = Bar()

# 3. 画图（数据、配置项）
bar.add_xaxis(['小米', '华为', '苹果', '三星', 'OPPO', '一加', '荣耀'])
bar.add_yaxis('手机品牌销量', [30, 20, 60, 16, 17, 18, 10])

# 4. 渲染图形
bar.render("第一个pyecharts图表.html")
```

**pyecharts 所有方法都支持链式调用**

```python
from pyecharts.charts import Bar

bar = (
    Bar()
    .add_xaxis(['小米', '华为', '苹果', '三星', 'OPPO', '一加', '荣耀'])
    .add_yaxis('手机品牌销量', [30, 20, 60, 16, 17, 18, 100])
)
bar.render("第一个pyecharts图表_链式调用.html")
```

配置项的简单使用

**在 pyecharts 中，一切皆配置 （options）**，掌握了配置，就学会了如何使用 pyecharts 去绘图

```python
from pyecharts import options as opts

bar = (
    Bar()
    .add_xaxis(['小米', '华为', '苹果', '三星', 'OPPO', '一加', '荣耀'])
    .add_yaxis('商家A', [30, 20, 60, 16, 17, 18, 10])
    .add_yaxis('商家B', [55, 12, 88, 56, 35, 45, 80])
    .add_yaxis('商家C', [55, 12, 88, 56, 35, 45, 175])

    # 全局配置项 set_global_opts（）
    .set_global_opts(
        # 标题配置项  
        title_opts=opts.TitleOpts(
            is_show=True,
            title='百度一下', 
            title_link='https://www.baidu.com/',
            subtitle='必应Bing',
            subtitle_link='https://cn.bing.com/',
            pos_left='left',
        )
    )
)
bar.render('配置项的简单使用.html')
```

## 全局配置项
全局配置项用于**设置图表的样式和布局**，包括标题、图例、提示框、坐标轴等。

`set_global_opts()`

问题：怎么知道如何配置里面的参数？答：去看官网的文档或者我给出的markdown文档。

常用的全局配置项如下：

1. 初始化配置 InitOpts
2. 标题配置 TitleOpts
3. 图例配置 LegendOpts
4. 工具箱配置 ToolboxOpts
5. 视觉映射配置 VisualMapOpts
6. 提示框配置 TooltipOpts
7. 缩放配置 DataZoomOpts

```python
from pyecharts.globals import ThemeType
from pyecharts import options as opts
from pyecharts.globals import CurrentConfig

CurrentConfig.ONLINE_HOST = "./"  # 切换到国内/更稳定的 CDN

from pyecharts import options as opts

bar = (
    Bar()
    .add_xaxis(['小米', '华为', '苹果', '三星', 'OPPO', '一加', '荣耀'])
    .add_yaxis('商家A', [30, 20, 60, 16, 17, 18, 10])
    .add_yaxis('商家B', [55, 12, 88, 56, 35, 45, 80])
    .add_yaxis('商家C', [55, 12, 88, 56, 35, 45, 175])

    # 全局配置项
    .set_global_opts(
        # 1.标题配置项 TitleOpts 
        title_opts=opts.TitleOpts(
            is_show=True,
            title='百度一下', 
            title_link='https://www.baidu.com/',
            subtitle='必应Bing',
            subtitle_link='https://cn.bing.com/',
            pos_left='left',
        ),
        # 2.图例配置项 LegendOpts
        legend_opts=opts.LegendOpts(
            is_show=True,
            type_='scroll',
            pos_left='15%',
            pos_top='15%',
            orient='vertical',
        ), 
        # 3.工具箱配置项 ToolboxOpts
        toolbox_opts=opts.ToolboxOpts(
            is_show=True,
            orient= "horizontal",
        ), 
        # 4.视觉映射 VisualMapOpts
        visualmap_opts=opts.VisualMapOpts(
            is_show= True,
        ),
        # 5.提示框配置 TooltipOpts
        tooltip_opts=opts.TooltipOpts(
            is_show=True,
        ),
        # 6.缩放配置 DataZoomOpts
        datazoom_opts=opts.DataZoomOpts(
            is_show=True,
            type_="slider",
            orient="vertical",
        )
    )
)
bar.render('全局配置项.html')
```

## 系列配置项
系列配置项用于**设置图表**中各个数据系列的属性，包括数据系列名称、数据、样式等。系列配置项对每个系列起作用，可用于个性化设置不同系列的样式和数据。
`set_series_opts()`


在线中文文档：https://www.w3cschool.cn/pyecharts/pyecharts-8ewr3q2e.html


1. 文本样式配置项 TextStyleOpts
2. ItemStyleOpts 图元样式配置项
3. 标签配置项 LabelOpts
4. 线样式配置项 LineStyleOpts
5. 标记点配置项 MarkPointOpts
6. 标记线配置项 MarkLineOpts
7. 标记区域配置项 MarkAreaOpts
8. 涟漪特效配置项 EffectOpts

```python
from pyecharts.globals import ThemeType
from pyecharts import options as opts

bar = (
    Bar()
    .add_xaxis(['小米', '华为', '苹果', '三星', 'OPPO', '一加', '荣耀'])
    .add_yaxis('商家A', [30, 20, 60, 16, 17, 18, 10])
    .add_yaxis('商家B', [55, 12, 88, 56, 35, 45, 80])
    .add_yaxis('商家C', [55, 12, 88, 56, 35, 45, 175])

    # 系列配置项
    .set_series_opts(
        itemstyle_opts=opts.ItemStyleOpts(
            color='RGBA(0, 255, 0, 0.5)'
        ),

        label_opts=opts.LabelOpts(
            is_show=True,
            position="top",
            color='red',
            font_size=15,
            font_style='oblique',  
            font_weight='bolder',  # 加粗
            font_family='Microsoft YaHei'
        ),
        linestyle_opts=opts.LineStyleOpts(
            is_show=True,
        ),
        splitline_opts=opts.SplitLineOpts(
            is_show=True,
        ),
    )
)
bar.render('系列配置项.html')
```

```python
# 任务一：调整上述组合图表，使得不会重叠，保证显示清楚

from pyecharts import options as opts
from pyecharts.charts import Bar, Grid, Line
from pyecharts.faker import Faker

# 创建柱状图
bar = (
    Bar()
    .add_xaxis(Faker.choose())
    .add_yaxis("商家A", Faker.values())
    .add_yaxis("商家B", Faker.values())
    .set_global_opts(
        title_opts=opts.TitleOpts(title="Grid-Bar"),
        legend_opts=opts.LegendOpts(pos_top="8%")
    )
)

# 创建折线图
line = (
    Line()
    .add_xaxis(Faker.choose())
    .add_yaxis("商家A", Faker.values())
    .add_yaxis("商家B", Faker.values())
    .set_global_opts(
        title_opts=opts.TitleOpts(title="Grid-Line", pos_top="52%"),
        legend_opts=opts.LegendOpts(pos_top="52%")
    )
)

# 使用Grid布局，合理设置上下间距
grid = (
    Grid(init_opts=opts.InitOpts(width="100%", height="800px"))
    .add(bar, grid_opts=opts.GridOpts(pos_top="15%", pos_bottom="58%"))
    .add(line, grid_opts=opts.GridOpts(pos_top="60%", pos_bottom="5%"))
    .render("8_组合图表_调整后.html")
)
```

```python
# Matplotlib 和 Pyecharts 的区别与优势对比

# Matplotlib的特点：
# 1. 是一个老牌的Python绘图库，功能非常全面
# 2. 主要生成静态图片，适合出版和学术文档
# 3. 语法较为底层，灵活性高但学习曲线相对陡峭
# 4. 对数据的控制精度高，可精细调整每个元素
# 5. 与NumPy、Pandas等科学计算库结合紧密
# 6. 适合制作学术图表、数据探索和科学可视化

# Pyecharts的特点：
# 1. 基于ECharts的Python接口，生成交互式HTML图表
# 2. 图表样式美观，有丰富的主题和交互功能
# 3. 链式调用的API设计，代码简洁易读
# 4. 支持多种图表类型，包括桑基图、地图、词云等特殊图表
# 5. 适合制作网页展示的可视化、交互式数据分析报告
# 6. 配置方便，支持全局配置和系列配置分离

# 主要区别：
# - 输出形式：Matplotlib是静态图片，Pyecharts是交互式HTML
# - 交互性：Pyecharts支持鼠标悬停、缩放、点击等交互
# - 学习曲线：Pyecharts的API设计更现代易用
# - 应用场景：Matplotlib适合学术出版，Pyecharts适合网页展示
```
